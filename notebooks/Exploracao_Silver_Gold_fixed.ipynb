# Exploração — Camadas Silver e Gold

Algumas linhas de cada tabela em `data/silver/` e `data/gold/`, só pra visualizar rapidamente o que tem em cada uma. Usa DuckDB (`LIMIT`) em vez de pandas puro porque algumas tabelas silver são grandes (`cnes_pf` ~1.1GB, `sia` ~130MB) — `LIMIT` evita carregar o arquivo inteiro na memória só pra ver 5 linhas.

```python
from pathlib import Path

import duckdb
import pandas as pd

from internasus.config import PROJ_ROOT, SILVER_DATA_DIR as SILVER_DIR, GOLD_DATA_DIR as GOLD_DIR

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)

con = duckdb.connect()


def amostra(caminho: Path, n: int = 5) -> pd.DataFrame:
    """Lê as primeiras n linhas de um parquet + conta o total de linhas, sem carregar o arquivo inteiro."""
    total = con.execute(f"SELECT COUNT(*) FROM read_parquet('{caminho.as_posix()}')").fetchone()[0]
    print(f"{caminho.relative_to(PROJ_ROOT)} -> {total:,} linhas")
    return con.execute(f"SELECT * FROM read_parquet('{caminho.as_posix()}') LIMIT {n}").df()
```
